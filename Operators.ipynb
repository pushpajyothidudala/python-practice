```python
#arithemetic operators
#+,-,*,/,//,%,**
#+ is used to add numbers
a=10+20
print(a)
#- is used for subtraction of two numbers
a=30-12
print(a)
#* is used for mul of two numbers
a=10*20
print(a)
#/ is used for floor division
a=20/10
print(a)
#// which gives result in int
a=20//10
print(a)
#% which gives remainder
a=10%6
print(a)
#**
a=10**2
print(a)
```

```python
#assignment operator and comparsion/relational operator
#+=,-=,=,*=,/=,//=,%=,**=
c=20;
c**=10;
print(c)
#>=,<=,<,>,!=,==
a=10;
b=10;
print(a==b)
```

```python
#logical operator
#not,and,or
a=10
b=20
if(not(a<=10)):
  print("sanju")
```

```python
#identity operator
#is,is not it will check both are indicating the same object,if yes it return true,or else false
a=10
b=10
print(a is b)
print(a is not b)
l=[1,2,3]
l2=[1,2,3]
print(l is l2)
print(l is not l2)
```

```python
#memebership operator-which is used to check whether a value is found in sequence or not
#in,not in
a="sanju"
print('ju' in a)
a={10:20,30:40}
print((10,20) in a.items())
```
