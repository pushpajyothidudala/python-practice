```python
#tuple is a list category data type,which is used store all datatype values either same or different
#tuple is similar to list,but tuple is immutable while list is immutable
#we cant the tuple values,they are constant or fixed values
v=(10,20,40);
#v[0]=1;TypeError: 'tuple' object does not support item assignment
#we can create two types of tuple object
#1.empty tuple
#empty-which doesnot contain any value and whose is length 0
#syntax-1:varname=();
t=();
print(t)
#syntax2:varname=tuple()
t=tuple()
print(t)
#non-empty tuple-contains data and whose length is geater than "0"
#syntax1:varname=(val1,val2,...valn)
t=(10,20,30,40,50)
print(t);
#syntax2:varname=tuple(Object)-object have to iterable
#t=tuple(10);TypeError: 'int' object is not iterable
t=tuple("sanju");#it calls the tuple constuctor and the object have to be iterable
print(t)
a=(10)
print(a,type(a))
#above we will get int ,it considers as a int,() does two operations in python
#one is used in bodmas operators,and second in tuple so when we use in tuple to avoid confusion we will use(,)
a=(10,)
print(a,type(a))
#syntax 3:obejct, here object is non iterable-,makes it tuple
t=10,
print(t,type(t))
```

```python
#predefined functions in the tuple
#count() is used to count the values
t=10,20,30,40,50,10,10,10
print(t.count(10))
#index() is used to the value of the index
print(t.index(10))
#if we want all the index then use enumerator
for i in enumerate(t):
  if(i[1]==10):
    print(i[0]);
#tuple wont allow reverse function,in dict we use tuple as key and if the key is reversed how can we get value
#tuple wont allow copy(),what is the use of copy when i cant perform any modifications
#we cant peform sort(),operation on the tuple,
#if we wnat to sort the immutable object then we wil sorted(),it will take the object and it will sort it and it never places it in the same object
#we need to store it in variable,it will return in list
t1=sorted(t)
print(tuple(t1))
t2=sorted("sanju")
print(tuple(t2))
```
