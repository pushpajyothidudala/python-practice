```python
#for variable_name in sequence
l=[1,2,3,4]
for s in l:
  print(s,end=" ")
print()
t=(1,2,3,4)
for s in t:
  print(s,end=" ")
print()
s={1,2,3,4}
for p in s:
  print(p,end=" ")

print()
d={1:2,3:4,5:6}
for k in d:
  print(k)
print()
for k in d.values():
  print(k,end=" ")
print()
for k,v,p,l in [(1,2,3,4)]:
  print(k," ",v," ",p," ",l)
print()

```

```python
#sum of all n numbers
y=eval(input("enter the number"))
sum=0
for i in range(y+1):
  sum=sum+i;
print(sum)
```

```python
#Read a word and a letter. Print how many times the letter appears in the word.
y=input("enter the word")
count=0;
for ch in y:
  if(ch=='a'):
    count=count+1;
print(count)

```

```python
#Using nums = [1, 2, 3, 4, 5, 6], print the sum of only the even numbers.
l=[1,2,3,4,5,6]
sum=0;
for i in range(len(l)):
  if(l[i]%2==0):
    print(l[i],i)
    sum=sum+l[i]
print(sum)
```

```python
#Read n and print a star triangle that gets smaller each row.
y=eval(input("enter the number"))
for i in range(y,0,-1):
  for j in range(i):
    print("*",end=" ")
  print()
```
