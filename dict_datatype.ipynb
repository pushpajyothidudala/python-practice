```python
#dict data type is used store key_value pair in an unordered way
#key must be hashable so key must be int,float,bool,str,tuple
#value can be of any datatype
#there are two types dict datatype availble
#empty_dict:which does not contain any elements and whose length is zero
#syntax:{} or dict()
d={}
print(d)
#non_empty_dict:which does contain elements and whose length is non-zero
#syntax:obj={key:value1,key:value2,..} or dict([()()])
d={"person1":"sanju","person2":"nani","person3":"prabhas"}
print(d)
#for syntax dict((outer(inner)))
#outer can be any iterable data type
#inner must two values in list,tuple,in dict
d=dict([{1,2},{2,3}])
print(d)
d=dict(((1,2),(2,3)))
print(d)
d=dict([("x",[1,2])])
print(d)
d=dict({(1,2),(2,3)})
print(d)
d=dict([{1:2,3:4}]) #it will take two values of key
print(d)
```

```python
#predefined funtcions in a dict
#dict.items():it will give you key and value pair in dict_item class
#these dict_items class elements we cant access through index
d={"person1":"sanju","person2":"nani","person3":"prabhas"}
print(d.items())
print(type(d.items()))
#dict.keys() is used to return a list of key
print(d.keys())
print(type(d.keys()))
#dict.values() is used to return a list of values
print(d.values())
print(type(d.values()))
```

```python
#adding items in a list
d={"person1":"sanju","person2":"nani","person3":"prabhas"}
d['person4']="chandu" # it will check whethher person 4 is thre or not ,if there it will update the value if not it will create a key and value pair
print(d)
#option2:to add multiple values we will use dict.update()
d.update([[1,2],[2,3]]) # update will go throught as key,value
print(d)
d.update({"person5":"sree","person6":"ram"})
print(d)
#to del the elements
del d["person6"]
print(d)
d.pop("person5")
print(d)
#to access specific value of a key
p=d['person1']
print(p[0])
p=d.get('person2')
print(p)
#to update the value
d['person1']='san'
print(d)
#to replace the key
d.update({'person':d.pop('person2')})
print(d)
```
