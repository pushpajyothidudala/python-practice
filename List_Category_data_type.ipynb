```python
#list is a predefined data type which is used to store list of values(same or different type) in a list
lst=[10,20.3,2+3j,"True",True]
print(lst);
#we can perform both indexing and slicing operations on list
a=lst[0];
print(a,type(a));
a=lst[0:3]
print(a,type(a))
lst[0:3]=[10,20,30];
print(a,lst[0:3]);
#on object of list is mutuable it is allowing to perform changes in the list in the original address
```

```python
#in python we have 2 types of list one is empty list and other is non empty list
#empty list:list which doesnot contain any data and whose length is zero
#syntax-1:lst=[];
lst=[];
print(lst);
#syntax-2:lst=list()-we are typecasting to list with zero objects
lst=list();
print(lst)
#non empty list:which does contain data and whose length is greater than zero
#syntax1:listobj=[val,val2,...valn];
lst=[10,20,30];
print(lst);
#syntax2:listobj=list(object)-object must be iterable object-we take any object we just converting it to list
lst=list("10");
print(lst);
#syntax3:listobj=list([object])-object is non iterable,convert them into iterable object in list and then type cast
lst=list([10]);
print(lst);
```

```python
#predefined functions in the list
#append()-which is used to append the items in the list at the end and it takes one argument
lst=[10,20,30,40];
lst.append(50);
print(lst)
lst.append([10,20,30]);
print(lst);
#if you want to them as sepearte elements then we have to use extend();
#extend()-which is used to a object to the end and it takes one argument and object have to iterable
lst.extend("sanju");
print(lst);
lst.extend({"a":"Sanju"});
print(lst);
#if you want to add one obj go to append,if we want to add multiple obj go to extend()
#both append and extend adds the obj at the end of the list,if you want to add at particular obj then we use insert();
#insert(index,value)-which is used to insert at particular position
lst=[10,20,30,40];
lst.insert(1,"sanju");
print(lst)
lst.insert(300,"chandu");
print(lst)
lst.insert(-100,"nani")
print(lst)
```

```python
#index(value)-if you want to find the particular index element and it will give indexerror if we give wwrong index
lst=[10,20,30,40,50,60]
print(lst[4])
a=lst[4]
print(a)
#if we want know all the index which 10 as a value then we use enumerator and range() function
#range(values)
lst=[10,20,30,10,40,20,10,30,10,50,10,70]
for i in range(len(lst)):
  if(lst[i]==20):
    print(i)
#range function works only when we have len() and indexing is possible if both are not possible then we cant use it ,we will go for enumerator
#enumerate()  takes the object convert each index elemnt value and its index and stores its value in the tuple
for i in enumerate(lst):
  print(i,type(i))

s="sanju";
for i in enumerate(s):
  if(i[1]=="s"):
    print(i,"prabbhas")
#count()-used for counting the number of occurences of the given elements,if element is not there it will zero as ans
a=lst.count(10)
print(a)
b=lst.count(100)
print(b)
```

```python
#copy(),there are two type of copy function present in python,one is shallow copy and another one is deep copy
#copy(),shallow copy is impleemented by using copy() function
#inital content is same,the memory address of both object is different and the modifications of one object is independent to other
lst=[10,20,30,40,50,60]
lst1=lst.copy();
lst1.append(10);
print(lst,lst1);
#deep copy is implemented by using assignment operator
#initial content is same,the memory address is same and modifications are dependent
lst=[10,20,30,40,50,60]
lst1=lst;
lst1.append(10);
print(lst,lst1);
```

```python
#reverse() function is used to reverse the values of the given list and returns none type
lst=[10,20,30,40,50]
lst1=lst.reverse();
print(lst1)
print(lst)
#if we want to place the reversed list in the new variable then use
lst1=lst[-1::-1];
print(lst1);
#or we can used reversed() built-in function
lst2=reversed([10,20,30]);
for x in lst2:
  print(x)
```

```python
#sort()-sort function is used to sort the data in ascending and descending order but the data must be same type
#syntax1:listobj.sort()-sort the data in the ascending order
lst=[100,90,80,70,60];
lst.sort()
print(lst)
#syntax2:listobj.sort(reversed=True)-sort the data in the descending order
lst.sort(reverse=True)
print(lst)
```
