```python
#there are several predefined string methods
#str.upper()-which is used to convert every character in the string to upper case
s="sanju"
print(s.upper())
#str.lower()-which is used to convert every charcter in the string to lower case
s="SANJU"
print(s.lower())
#str.swapcase()-which is used to convert upper to lower and lower to upper case
s="SAnju"
print(s.swapcase())
#str.captalize()-which is used to  convert every first letter in the string to caps
s="sanju is very intelligent"
print(s.capitalize())
#str.title()-which is used to  convert every first word letter in the string to caps
s="sanju is very intelligent"
print(s.title())
#str.strip()-which is used to trim the spaces in the left and right end of the string
s=" Sanju "
print(s.strip())
#str.lstrip()-which is used to trim the spaces in the left  end of the string
s=" Sanju "
print(s.lstrip())
#str.rstrip()-which is used to trim the spaces in the  right end of the string
s=" Sanju "
print(s.rstrip())
#str.replace()-which is used to replace every occurrence with the given string
s="MISSISSIPI"
print(s.replace("I","j"))
#str.split(Delimiter)-if no delimiter is given it will consider white space
s="sanju#is#a#good#girl"
p=s.split("#")
print(p,type(p))
#seperator.join(iterbale)-which is used to join a list of strings into a single thing,using a specified delimiter
s="hi"
pr="::".join(p)
print(pr,type(pr))
#str.index()-which give the index of the first ocuurence of the given character,if there is no character then it raises error
s="MISSISSIPI"
print(s.index('I'))
#str.find()-which gives the index of the first ocuurence,if there is no charcter it will give -1
print(s.find("M"))
#str.count()-which counts the given occurence
print(s.count("SS"))
```

```python
#str.startswith()-it checks it will start that charcter or not
s="sanju123"
print(s.startswith("s"))
#str.endswith()-it checks it will end that charcter or not
print(s.endswith("p"))
#str.isalnum()-whether a string contains only letters and digits no special character is allowed
print(s.isalnum())
#str.isalpha():only alphabets
print(s.isalpha())
#str.isdigit()-only digits
print(s.isdigit())
#str.islower()-only lower case
print(s.islower())
#str.isupper()-only upper case
print(s.isupper())
#str.isspace()-only contains whitespace
p=" "
print(p.isspace())
print(s.isspace())
#str.istitle()-checks every word first character is caps or not
print(s.istitle())
#str.removeprefix()-it removes given prefix character
print(s.removeprefix("sa"))
#str.removesuffix()-it removes given suffix character
print(s.removesuffix("123"))
```
