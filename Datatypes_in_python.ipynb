```python
#variables
#In python we dont have to declare the type of the variables ,variables  type declaration happens automatically
#the equal sign is used to assign values to variables
x=1;
print(x);
print(type(x));
y="sanju";
print(y);
print(type(y));
```

```python
#fundtamental data types
#int,float,bool,complex
#int is used to define interger values
x=1;
print(x, type(x));
#float is used to define floating values
y=12.75;
print(y,type(y));
#bool is used to define true or false values
z=True;
print(z,type(z));
#complex is used to define both real and imaginary numbers,both real and imaginary are floating points numbers.
a=10+25j;
print(a,type(a));
#if we want to get real part then
print(a.real,"real");
#if we want to get img part then
print(a.imag,"img");
```

```python
#sequence category data types
#sequence category data types are used to store sequence of values in a single object
#they are 4 type of sequence category data types
#str,bytes,bytearray,range
#str is used to store mutliple character in a single object,in python both characters and sequence of multiple character both are handled by strings only,
#str can be represented by single or double quotes
s="a";
print(s,type(s));
s='python';
print(s,type(s));
#if you want to represent the data in the multiple lines then use """ """ or ''' '''
p="""sanju
is a
good girl""";
print(p,type(p));
```

```python
#operations on str
#indexing,slicing
#indexing is used to get a corresponding character in the given index,if we give invalid index then we will get index error
s="python";
print(s[3]);
print(s[True]);
print(s[-True])
#print(s[12]);
#indexing will get one charcter,but if we want multiple character from the single object then we will use slicing
#to perform slicing operations,we have 5 syntaxs ,they are
#syntax-1:strobj[begin:end] we will get from begin to end-1 characters
p=s[3:5];
print(p);
p=s[0:6];
print(p);
#syntax-2:strobj[begin:] if we dont know the end value or if we want to take until the end of the string then we dont need to specify the end
p=s[2:];
print(p);
#syntax-3:strobj[:end] if we want to take from 0 to end then we can use this
p=s[:4];
print(p);
#syntax-4:str[:] if you want to take the entire object without specifying end or begin then use this one
p=s[:];
print(p);
#in the above all syntax all we hhave to follow is begin<end or else we will get '' if step is positive





```

```python
#we can present indexing in negative way also ,if we use negative indexing then syntax1:
s="python";
p=s[-6:-2]; # -6+6 is 0 begin is 0 and end is 6-2 is 4 then 4-1 is 3 then end is 3 0 to 3
print(p);
p=s[-6:0]; #When we are default step as "1" then begin<end orelse we will get space or ' ',in this seneraio obviously we will take negative numbers as len(str)+our number then -6+6 is 0 and 0 is positive then 0<0 is ' '
print(p);
p=s[-4:6];
print(p);
p=s[2:-2];
print(p);
p=s[2:-4]; # begin is 2 and end is 6-4 is 2 then 2-1 is 1 then 2 to 1 and begin>end ,so we will get space
print(p);
p=s[-1:0]; #bgin is 6-1 is 5 and end is 0 then 5 to -1 begin>end we will get space
print(p);
p=s[-1:] # begin is 5 and end is 6-1 is 5
print(p);
p=s[-4:]
print(p);# begin 6-4 is 2 and end is 6-1 is 5 then 2 to 5
p=s[15:];# begin is 15 and end is 5 where end<begin so space;
p=s[-len("sanju"):-1];# begin is -5 so 6-5 is 1 and end is 6-1 is 5-1 is 4 from 1 to 4
print(p);
#in the above all the 4 syntax we should always remember that begin<end.


```

```python
#now we will look into syntax 5
#syntax 5:strobj[begin:end:step] have 5 rules
#rule1:here the value of begin,end and step can be either _ve or +ve
#rule2:if the value of the step is +ve then pvm gives the range of chars from begin to end-1 index in the forward direction where begin<end
s="python";
p=s[-5:-1:1]
print(p);
p=s[0:6:2];
print(p);
p=s[-1:-6:1];
print(p);
#rule 3:if the value of step is -ve, then pvm gives the range of chard in backward direction from begin to endindex+1 provided begin>end otherwise we will get " "as result
p=s[6:3:-1];
print(p);
p=s[::-1]; # begin is 5 and end is end of the string begin>end so we will get reverse string
print(p);
#rule 4: when you take the step as +ve one then if we specify endindex as zero then we will get ' ' as result
p=s[0:0:1];
print(p);
#rule 4: when you take the step as -ve one then if we specify endindex as -1 then we will get ' ' as result
p=s[:-1:-1];
print(p);
```
